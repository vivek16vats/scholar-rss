```python
from scholarly import scholarly
from feedgen.feed import FeedGenerator
from datetime import datetime
import os

# ============================================
# Google Scholar User ID
# ============================================

SCHOLAR_USER_ID = "TKbYqt0AAAAJ"

# ============================================
# Fetch Scholar Profile
# ============================================

print("Fetching Google Scholar profile...")

author = scholarly.search_author_id(SCHOLAR_USER_ID)

author = scholarly.fill(
    author,
    sections=["publications"]
)

publications = author.get("publications", [])

print(f"Found {len(publications)} publications")

# ============================================
# Create RSS Feed
# ============================================

fg = FeedGenerator()

fg.title(f"{author['name']} - Google Scholar Publications")
fg.link(href=f"https://scholar.google.com/citations?user={SCHOLAR_USER_ID}")
fg.description(
    f"RSS feed for scientific publications of {author['name']}"
)
fg.language("en")

# ============================================
# Add Publications
# ============================================

for pub in publications:

    try:
        pub_filled = scholarly.fill(pub)

        bib = pub_filled.get("bib", {})

        title = bib.get("title", "No Title")

        year = bib.get("pub_year", "Unknown Year")

        authors = bib.get("author", "")

        abstract = bib.get("abstract", "")

        pub_url = pub_filled.get(
            "pub_url",
            f"https://scholar.google.com/citations?user={SCHOLAR_USER_ID}"
        )

        entry = fg.add_entry()

        entry.title(title)

        entry.link(href=pub_url)

        entry.description(
            f"""
            Authors: {authors}<br>
            Year: {year}<br><br>
            {abstract}
            """
        )

        entry.guid(pub_url, permalink=True)

        try:
            pub_date = datetime(int(year), 1, 1)
            entry.pubDate(pub_date)
        except:
            pass

        print(f"Added: {title}")

    except Exception as e:
        print(f"Skipping publication due to error: {e}")

# ============================================
# Save RSS Feed
# ============================================

output_file = "scholar_feed.xml"

fg.rss_file(output_file)

print(f"\nRSS feed generated successfully!")
print(f"Saved as: {os.path.abspath(output_file)}")
```

```python


from scholarly import scholarly
from feedgen.feed import FeedGenerator

AUTHOR_ID = "TKbYqt0AAAAJ"

author = scholarly.fill(
    scholarly.search_author_id(AUTHOR_ID)
)

fg = FeedGenerator()
fg.title(author["name"] + " Publications")
fg.link(href=author["url_picture"] if "url_picture" in author else "")
fg.description("Google Scholar Publications")

for pub in author["publications"]:
    try:
        p = scholarly.fill(pub)

        bib = p["bib"]

        fe = fg.add_entry()
        fe.title(bib.get("title", ""))

        summary = f"""
        Authors: {bib.get('author', '')}
        Venue: {bib.get('venue', '')}
        Year: {bib.get('pub_year', '')}
        """

        fe.description(summary)

    except Exception:
        pass

fg.rss_file("scholar.xml")
```

```python
import requests
import json
import time

API_KEY = "fedd0e6f5775babafdf781bb455bec1461ab73025e105f7f910e70aa2b852b4d"
AUTHOR_ID = "TKbYqt0AAAAJ"

BASE_URL = "https://serpapi.com/search.json"

all_articles = []
seen_titles = set()

author_info = None
citation_info = None

start = 0
page = 1

while True:

    params = {
        "engine": "google_scholar_author",
        "author_id": AUTHOR_ID,
        "hl": "en",
        "start": start,
        "sort": "pubdate",
        "api_key": API_KEY
    }

    print(f"Fetching page {page} (start={start})")

    response = requests.get(BASE_URL, params=params)

    if response.status_code != 200:
        print(f"Error {response.status_code}")
        print(response.text)
        break

    data = response.json()

    # Save author details once
    if author_info is None:
        author_info = data.get("author", {})
        citation_info = data.get("cited_by", {})

    articles = data.get("articles", [])

    # Deduplicate
    for article in articles:

        title = article.get("title", "").strip()

        if title and title not in seen_titles:
            seen_titles.add(title)
            all_articles.append(article)

    print(f"Collected {len(articles)} articles")
    print(f"Running total: {len(all_articles)}")

    # Stop immediately if no next page
    pagination = data.get("serpapi_pagination", {})

    if "next" not in pagination:
        print("No more pages found.")
        break

    # Next page exists -> make one more call
    start += len(articles)
    page += 1

    # Optional: avoid hitting rate limits
    time.sleep(1)

# Final combined JSON
final_result = {
    "author": author_info,
    "cited_by": citation_info,
    "total_articles": len(all_articles),
    "articles": all_articles
}

# Save complete result
with open("scholar_complete.json", "w", encoding="utf-8") as f:
    json.dump(
        final_result,
        f,
        indent=2,
        ensure_ascii=False
    )

print(f"\nTotal unique articles: {len(all_articles)}")
print("Saved to scholar_complete.json")
```
